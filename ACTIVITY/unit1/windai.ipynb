```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Wspd"].dropna(),
    bins=40,
    kde=True
)

plt.title("Distribution of Wind Speed")
plt.xlabel("Wind Speed")
plt.ylabel("Frequency")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df["Wspd"].dropna()
)

plt.title("Box Plot of Wind Speed")
plt.ylabel("Wind Speed")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    y=df["Patv"].dropna()
)

plt.title("Box Plot of Active Power")
plt.ylabel("Active Power (Patv)")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    data=df.sample(min(10000, len(df))),
    x="Wspd",
    y="Patv",
    alpha=0.5
)

plt.title("Wind Speed vs Active Power")
plt.xlabel("Wind Speed")
plt.ylabel("Active Power")
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 8))

correlation = numeric_df.corr()

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Heatmap of Numerical Features")
plt.tight_layout()
plt.show()
```

```python
column = "Patv"

Q1 = df[column].quantile(0.25)
median = df[column].quantile(0.50)
Q3 = df[column].quantile(0.75)

mean = df[column].mean()

P10 = df[column].quantile(0.10)
P90 = df[column].quantile(0.90)

print("Mean:", mean)
print("Median:", median)
print("Q1:", Q1)
print("Q3:", Q3)
print("10th Percentile:", P10)
print("90th Percentile:", P90)
```
